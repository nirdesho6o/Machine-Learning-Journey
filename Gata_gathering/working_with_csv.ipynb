### Opening CSV from URL 

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/MainakRepositor/Datasets/master/Air%20Quality/real_2013_air.csv"
#replace github.com with raw.githubusercontent.com and remove /blob/ segment as it is internal instruction by git for HTML on webpage and not on raw data 
req = requests.get(url)
data = StringIO(req.text)

pd.read_csv(data)
```

## Sep Parameter

```python
#TSV are tab seperated values . to read these using pandas we write 
pd.read_csv('movie_titles_metadata.tsv',sep='\t')

#by default for csv files the sep is ','
```
