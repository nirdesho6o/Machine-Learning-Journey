## One hot encoding is done on nominal data to convert categorical to numeric 

## best way to understand is to convert string to vector of 1 and 0. 

## dummy variable trap is concept where if n categories are there we keep n-1 columns bcoz of relation bw input categories

### if the categores are very high we can merge the least frequestly occuring ones as OTHERS

```python
import numpy as np 
import pandas as pd
```

```python
df=pd.read_csv('cars.csv')
df.head()
```

# OneHotEncoding using pandas

```python
pd.get_dummies(df,columns=['fuel','owner'],dtype=int) #by default pandas gives boolean so dtype=int is mentioned
#pandas does not remember the order of columns so it is avoided-> scikit is prefered
```

# K-1 OneHotEncoding

```python
pd.get_dummies(df,columns=['fuel','owner'],drop_first=True,dtype=int) #dummy variable trap is concept where if n categories are there we keep n-1 columns bcoz of relation bw input categories¶
```

#  OneHotEncoding Using Scikit Learn 

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.iloc[:,0:4],df.iloc[:,-1],test_size=0.2,random_state=2)
```

```python
X_test.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(drop='first',sparse_output=False,dtype=np.int32)
```

```python
X_train_new = ohe.fit_transform(X_train[['fuel','owner']])
X_test_new = ohe.transform(X_test[['fuel','owner']])
X_train_new.shape
```

```python
np.hstack((X_train[['brand','km_driven']].values,X_train_new))
```

# OneHotEncoding with Top Categories -- the lowest freq are clubbed as uncommon

```python
counts = df['brand'].value_counts()
df['brand'].nunique()
threshold = 100
repl = counts[counts <= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(repl, 'uncommon')).sample(5)
```
