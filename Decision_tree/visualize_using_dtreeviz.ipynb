```python

import graphviz.backend as be
from sklearn import tree
```

```python

from sklearn.datasets import *
from dtreeviz.trees import *
from IPython.display import Image, display_svg, SVG
```

```python

clas = tree.DecisionTreeClassifier()  
iris = load_iris()

X_train = iris.data
y_train = iris.target
clas.fit(X_train, y_train)
```

```python

from sklearn.tree import plot_tree
plot_tree(clas)
```

```python
import dtreeviz


viz_model = dtreeviz.model(clas, 
                           X_train=X_train, 
                           y_train=y_train, 
                           feature_names=iris.feature_names, 
                           class_names=["setosa", "versicolor", "virginica"])
viz = viz_model.view(fontname="DejaVu Sans")
viz

```

```python
df_cars = pd.read_csv("auto-mpg.csv")

df_cars
```

```python
import dtreeviz
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

# 1. Prepare data and train model
X = df_cars[['weight', 'acceleration']]
y = df_cars['mpg']
dt = DecisionTreeRegressor(max_depth=3, criterion="absolute_error")
dt.fit(X, y)

# 2. Initialize the modern dtreeviz model adaptor
viz_model = dtreeviz.model(dt, 
                           X_train=X, 
                           y_train=y, 
                           feature_names=['Vehicle Weight', 'Horse Power'],
                           target_name='mpg')

# 3. Setup the Matplotlib figure for 3D
figsize = (6, 5)
fig = plt.figure(figsize=figsize)
ax = fig.add_subplot(111, projection='3d')

# 4. Use the model's built-in 3D feature space method
# This replaces the old rtree_feature_space function
viz_model.rtree_feature_space3D(features=['Vehicle Weight', 'Horse Power'],
                                fontsize=14,fontname='DejaVu Sans',
                                elev=20,
                                azim=25,
                                show={'splits', 'title'},
                                ax=ax)

plt.show()
```
